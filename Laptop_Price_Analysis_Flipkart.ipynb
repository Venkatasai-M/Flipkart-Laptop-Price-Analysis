## 14. Data Collection: Web Scraping Utilities

### 14.1 Imports, Session, and Request Headers


```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import numpy as np
import time
import random


# Reusable session for connection pooling across all requests
session = requests.Session()

# Realistic browser-like headers to reduce request blocking
HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/124.0.0.0 Safari/537.36"
    ),
    "Accept": (
        "text/html,application/xhtml+xml,application/xml;"
        "q=0.9,image/avif,image/webp,*/*;q=0.8"
    ),
    "Accept-Language": "en-US,en;q=0.9",
    "Connection": "keep-alive",
}

session.headers.update(HEADERS)
```

### 14.2 `get_soup()` — Fetch and Parse a Page

```python
def get_soup(url, timeout=10, retries=3, backoff_factor=1.5):
    '''Send a GET request to `url` and return a BeautifulSoup object.

    Retries on timeout/connection errors with exponential backoff.
    Returns None if all attempts fail.
    '''
    for attempt in range(1, retries + 1):
        try:
            response = session.get(url, timeout=timeout)
            response.raise_for_status()
            return BeautifulSoup(response.text, "html.parser")

        except requests.exceptions.Timeout:
            print(f"[Attempt {attempt}] Timeout while fetching: {url}")

        except requests.exceptions.HTTPError as http_err:
            print(f"[Attempt {attempt}] HTTP error for {url}: {http_err}")

        except requests.exceptions.ConnectionError:
            print(f"[Attempt {attempt}] Connection error while fetching: {url}")

        except requests.exceptions.RequestException as req_err:
            print(f"[Attempt {attempt}] Request failed for {url}: {req_err}")

        if attempt < retries:
            sleep_time = backoff_factor * attempt + random.uniform(0, 1)
            time.sleep(sleep_time)

    print(f"Failed to fetch {url} after {retries} attempts.")
    return None
```

### 14.3 `extract_product_cards()` — Locate Laptop Containers on a Page

```python
def extract_product_cards(soup):
    """
    Return all laptop product cards from the Flipkart search page.
    """
    if soup is None:
        return []

    return soup.find_all("a", class_="k7wcnx")
```

## 15. Data Collection: Extracting Laptop Details

### 15.1 `extract_laptop_details()` — Parse a Single Product Card


```python
import re

# -------------------------------
# Helper Functions
# -------------------------------

def clean_text(text):
    """Clean extracted text by removing extra spaces and newlines."""
    return re.sub(r"\s+", " ", text).strip() if text else None


def safe_get_text(element):
    """Safely return cleaned text from a BeautifulSoup element."""
    return clean_text(element.get_text()) if element else None


# -------------------------------
# Parse Laptop Specifications
# -------------------------------

def parse_specifications(card):
    specs = {
        "Processor": None,
        "RAM": None,
        "Storage": None,
        "Display Size": None,
        "Operating System": None,
        "Graphics Card": None,
        "Warranty": None,
    }

    spec_items = card.find_all("li", class_="DTBslk")

    for item in spec_items:
        text = clean_text(item.get_text())
        if not text:
            continue

        lower = text.lower()

        if specs["Processor"] is None and any(x in lower for x in ["processor", "core", "ryzen", "celeron", "pentium"]):
            specs["Processor"] = text

        elif specs["RAM"] is None and "ram" in lower:
            specs["RAM"] = text

        elif specs["Storage"] is None and any(x in lower for x in ["ssd", "hdd", "emmc"]):
            specs["Storage"] = text

        elif specs["Display Size"] is None and ("inch" in lower or "display" in lower):
            specs["Display Size"] = text

        elif specs["Operating System"] is None and any(
            x in lower for x in ["windows", "dos", "chrome", "mac", "operating system"]
        ):
            specs["Operating System"] = text

        elif specs["Graphics Card"] is None and any(
            x in lower for x in ["graphics", "nvidia", "amd", "intel iris", "geforce", "rtx"]
        ):
            specs["Graphics Card"] = text

        elif specs["Warranty"] is None and "warranty" in lower:
            specs["Warranty"] = text

    return specs


# -------------------------------
# Extract Laptop Details
# -------------------------------

def extract_laptop_details(card):
    try:
        # Product Name
        product_name = safe_get_text(card.find("div", class_="RG5Slk"))

        # Brand
        brand = product_name.split()[0] if product_name else None

        # Prices
        current_price = safe_get_text(card.find("div", class_="hZ3P6w"))
        original_price = safe_get_text(card.find("div", class_="kRYCnD"))

        # Discount
        discount = safe_get_text(card.find("div", class_="HQe8jr"))

        # Rating
        rating = safe_get_text(card.find("div", class_="MKiFS6"))

        # Ratings & Reviews
        number_of_ratings = safe_get_text(card.find("span", class_="PvbNMB"))

        # Product URL
        href = card.get("href")
        product_url = (
            "https://www.flipkart.com" + href
            if href else None
        )

        # Specifications
        specs = parse_specifications(card)

        laptop_details = {
            "Brand": brand,
            "Product Name": product_name,
            "Current Price": current_price,
            "Original Price": original_price,
            "Discount": discount,
            "Rating": rating,
            "Number of Ratings": number_of_ratings,
            "Processor": specs["Processor"],
            "RAM": specs["RAM"],
            "Storage": specs["Storage"],
            "Display Size": specs["Display Size"],
            "Operating System": specs["Operating System"],
            "Graphics Card": specs["Graphics Card"],
            "Warranty": specs["Warranty"],
            "Product URL": product_url,
        }

        return laptop_details

    except Exception as error:
        print(f"Error extracting laptop details: {error}")
        return None
```

```python
all_laptops = []

BASE_URL = "https://www.flipkart.com/search?q=laptops&sid=6bo,b5g"

TOTAL_PAGES = 20      # 20 pages × ~24 laptops ≈ 480 laptops

for page in range(1, TOTAL_PAGES + 1):

    print(f"Scraping Page {page}...")

    url = f"{BASE_URL}&page={page}"

    soup = get_soup(url)

    if soup is None:
        print(f"Could not fetch page {page}")
        continue

    cards = extract_product_cards(soup)

    print(f"Found {len(cards)} products")

    for card in cards:
        details = extract_laptop_details(card)
        if details:
            all_laptops.append(details)

    # polite delay
    time.sleep(random.uniform(2, 4))

print("\nScraping Completed!")
print("Total Products:", len(all_laptops))
```

```python
def simplify_processor(proc):
    proc = proc.lower()
    # Intel Core series
    if 'i3' in proc or 'core 3' in proc:
        return 'Intel i3'
    elif 'i5' in proc or 'core 5' in proc:
        return 'Intel i5'
    elif 'i7' in proc or 'core 7' in proc:
        return 'Intel i7'
    elif 'i9' in proc or 'core 9' in proc:
        return 'Intel i9'
    # AMD Ryzen series
    elif 'ryzen 3' in proc:
        return 'AMD Ryzen 3'
    elif 'ryzen 5' in proc:
        return 'AMD Ryzen 5'
    elif 'ryzen 7' in proc:
        return 'AMD Ryzen 7'
    elif 'ryzen 9' in proc:
        return 'AMD Ryzen 9'
    # Qualcomm Snapdragon
    elif 'snapdragon' in proc:
        return 'Qualcomm Snapdragon'
    else:
        return 'Other'

# Apply the function
df['Processor_Group'] = df['Processor'].apply(simplify_processor)

# Example plot: Brand vs Processor_Group
plt.figure(figsize=(12,6))
sns.countplot(x='Brand', hue='Processor_Group', data=df)
plt.title("Brand vs Processor Group (Simplified)")
plt.xticks(rotation=45)
plt.show()
```

```python

# 1. Key summary statistics
total_laptops = df.shape[0]
total_brands = df["Brand"].nunique()
avg_price = df["Current Price"].mean()
max_price_row = df.loc[df["Current Price"].idxmax()]
min_price_row = df.loc[df["Current Price"].idxmin()]
avg_rating = df["Rating"].mean()
most_common_brand = df["Brand"].mode()[0]
most_common_os = df["Operating System"].mode()[0]
most_common_ram = df["RAM"].mode()[0]
most_common_storage = df["Storage"].mode()[0]

print(f"Total Laptops: {total_laptops}")
print(f"Total Brands: {total_brands}")
print(f"Average Price: {avg_price:.2f}")
print(f"Highest Priced Laptop: {max_price_row['Product Name']} ({max_price_row['Current Price']})")
print(f"Lowest Priced Laptop: {min_price_row['Product Name']} ({min_price_row['Current Price']})")
print(f"Average Rating: {avg_rating:.2f}")
print(f"Most Common Brand: {most_common_brand}")
print(f"Most Common OS: {most_common_os}")
print(f"Most Common RAM: {most_common_ram} GB")
print(f"Most Common Storage: {most_common_storage} GB")
```
